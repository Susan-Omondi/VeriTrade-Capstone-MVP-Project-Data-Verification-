```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")

from google.colab import drive
drive.mount('/content/drive')
```

```python

df = pd.read_csv('/content/drive/MyDrive/mock_suppliers_cac_registry_3500_.csv')
df.head()
```

```python
# Let's take a look at out data to understand it.
df.info()
```

```python
# tax id is changed to sting because it's not suppose to be a float. Floats can cause precision issues and formatting errors
df['tax_id'] = df['tax_id'].astype(str)
print(df)
```

```python
# Date fix. Even though going through the data set, the date was good but for accuracy and safety levels, we'll fix it.
df["registration_date"] = pd.to_datetime(df["registration_date"], errors="coerce")
print(df)
```

```python
# What we are going to use to flag the business  #helper flags
# If date is missing or unreadable put True
# If date is okay put False

df["flag_invalid_registration_date"] = df["registration_date"].isnull()
print(df)
```

```python
from datetime import datetime

# Recently registered,  less than 2 years

today = pd.Timestamp(datetime.today())

df["flag_recent_registration"] = ((today - df["registration_date"]).dt.days < 730)
print(df.head())
```

```python
# Missing tax id
df["flag_missing_tax_id"] = df["tax_id"].isnull()
print(df)
```

```python
# New verification logic for the data- we are replacing the old logic

def verify_supplier(row):

    # REJECT rules
    if row["status"].lower() != "registered":
        return "REJECTED", "Business not registered"

    if pd.isnull(row["registration_date"]):
        return "REJECTED", "Missing or invalid registration date"

    # FLAG rules
    if row["flag_recent_registration"]:
        return "FLAGGED", "Recently registered business"

    if row["flag_missing_tax_id"]:
        return "FLAGGED", "Missing tax identification number"

    # VERIFIED
    return "VERIFIED", "Business passed basic verification"


```

```python
# Apply the logic to the dataset
df[["verification_status", "verification_reason"]] = df.apply(
    verify_supplier,
    axis=1,
    result_type="expand")
print(df.head())
```

```python
# Final column that we have in our dataset
df_mvp = df[["supplier_id", "business_name", "registration_number", "status", "registration_date",
             "business_type", "industry_category",  "state", "verification_status", "verification_reason"]]
print(df_mvp)
```
